```python

df = pd.read_csv("Titanic-Dataset.csv")
print("Dataset loaded successfully!")

# Exploration
print("\nFirst 5 rows:")
display(df.head())

print("\nDataset shape (rows, cols):", df.shape)
print("\nMissing values per column:")
display(df.isnull().sum())
```

```python
# 1. Filter the data
custom_query = df.loc[(df["Sex"] == "female") & (df["Pclass"] == 3) & (df["Fare"] > 20)]

# 2. Fetch answers
total_women = len(custom_query)
survivors = custom_query["Survived"].sum()

print(f"Total female passengers in 3rd class who paid > 20: {total_women}")
print(f"Number of survivors among them: {survivors}")

print("\nSample of these passengers:")
display(custom_query[["Name", "Age", "Fare", "Survived"]].head(5))
```

```python
print("Overall Survival Counts:")
display(df["Survived"].value_counts())

print("\nOverall Survival Percentages:")
display(df["Survived"].value_counts(normalize=True) * 100)
```
